```python
# ============================================================
# QUESTION 2 — Dominant Strategy Finder
# ============================================================

def dominant_strategy(matrix):

    rows = len(matrix)
    cols = len(matrix[0])

    # -----------------------------
    # Check dominant row for P1
    # -----------------------------
    p1_dominant = None

    for r1 in range(rows):

        is_dominant = True

        for r2 in range(rows):

            if r1 == r2:
                continue

            for c in range(cols):

                if matrix[r1][c][0] <= matrix[r2][c][0]:
                    is_dominant = False
                    break

            if not is_dominant:
                break

        if is_dominant:
            p1_dominant = r1
            break

    # -----------------------------
    # Check dominant column for P2
    # -----------------------------
    p2_dominant = None

    for c1 in range(cols):

        is_dominant = True

        for c2 in range(cols):

            if c1 == c2:
                continue

            for r in range(rows):

                if matrix[r][c1][1] <= matrix[r][c2][1]:
                    is_dominant = False
                    break

            if not is_dominant:
                break

        if is_dominant:
            p2_dominant = c1
            break

    # -----------------------------
    # Print Results
    # -----------------------------
    print("\nDominant Strategy Results")

    if p1_dominant is not None:
        print("P1 dominant row:", p1_dominant)
    else:
        print("P1 has no dominant strategy")

    if p2_dominant is not None:
        print("P2 dominant column:", p2_dominant)
    else:
        print("P2 has no dominant strategy")


# Example Test

matrix = [
    [(6, 4), (8, 1)],
    [(4, 6), (7, 3)]
]

dominant_strategy(matrix)
```

```python

# ============================================================
# QUESTION 3 — IESDS Solver
# ============================================================
\
def iesds(matrix):

    rows = list(range(len(matrix)))
    cols = list(range(len(matrix[0])))

    changed = True

    while changed:

        changed = False

        # -----------------------------------
        # Check strictly dominated rows
        # -----------------------------------
        remove_row = None

        for r1 in rows:

            for r2 in rows:

                if r1 == r2:
                    continue

                dominated = True

                for c in cols:

                    if matrix[r1][c][0] >= matrix[r2][c][0]:
                        dominated = False
                        break

                if dominated:
                    remove_row = r1
                    print(f"Remove Row {r1} (dominated by Row {r2})")
                    break

            if remove_row is not None:
                break

        if remove_row is not None:
            rows.remove(remove_row)
            changed = True
            continue

        # -----------------------------------
        # Check strictly dominated columns
        # -----------------------------------
        remove_col = None

        for c1 in cols:

            for c2 in cols:

                if c1 == c2:
                    continue

                dominated = True

                for r in rows:

                    if matrix[r][c1][1] >= matrix[r][c2][1]:
                        dominated = False
                        break

                if dominated:
                    remove_col = c1
                    print(f"Remove Column {c1} (dominated by Column {c2})")
                    break

            if remove_col is not None:
                break

        if remove_col is not None:
            cols.remove(remove_col)
            changed = True

    print("\nRemaining Strategies")
    print("Rows:", rows)
    print("Columns:", cols)


# Example Test

matrix = [
    [(3,3), (2,2), (1,1)],
    [(4,4), (3,3), (2,2)],
    [(5,5), (4,4), (3,3)]
]

iesds(matrix)
```

```python
# ============================================================
# QUESTION 4 — Pure Nash Equilibrium Finder
# ============================================================

def pure_nash(matrix):

    rows = len(matrix)
    cols = len(matrix[0])

    equilibria = []

    for r in range(rows):

        for c in range(cols):

            # ---------------------------
            # Check P1 best response
            # ---------------------------
            p1_best = True

            for rr in range(rows):

                if matrix[rr][c][0] > matrix[r][c][0]:
                    p1_best = False
                    break

            # ---------------------------
            # Check P2 best response
            # ---------------------------
            p2_best = True

            for cc in range(cols):

                if matrix[r][cc][1] > matrix[r][c][1]:
                    p2_best = False
                    break

            # ---------------------------
            # Nash Equilibrium
            # ---------------------------
            if p1_best and p2_best:
                equilibria.append((r, c))

    print("\nPure Nash Equilibria")

    if equilibria:
        for eq in equilibria:
            print(eq)
    else:
        print("No Pure Nash Equilibrium")


# Prisoner's Dilemma Example

pd = [
    [(-1,-1), (-3,0)],
    [(0,-3), (-2,-2)]
]

pure_nash(pd)
```

```python
# ============================================================
# QUESTION 5 — N-Player Nash Equilibrium Finder
# ============================================================

from itertools import product

def nash_n_player(strategies, payoffs):

    players = len(strategies)

    profiles = list(product(*strategies))

    equilibria = []

    for profile in profiles:

        is_nash = True

        # -----------------------------------
        # Check each player's deviation
        # -----------------------------------
        for p in range(players):

            current_payoff = payoffs[profile][p]

            for alt in strategies[p]:

                if alt == profile[p]:
                    continue

                new_profile = list(profile)
                new_profile[p] = alt
                new_profile = tuple(new_profile)

                new_payoff = payoffs[new_profile][p]

                if new_payoff > current_payoff:
                    is_nash = False
                    break

            if not is_nash:
                break

        if is_nash:
            equilibria.append(profile)

    # -----------------------------------
    # Print Result
    # -----------------------------------
    print("\nNash Equilibria")

    if equilibria:
        for eq in equilibria:
            print(eq)
    else:
        print("No Nash Equilibrium")


# 3-Player Example

strategies = [
    ['C', 'D'],
    ['C', 'D'],
    ['C', 'D']
]

payoffs = {

    ('C','C','C'): [4,4,4],
    ('C','C','D'): [1,1,6],
    ('C','D','C'): [1,6,1],
    ('C','D','D'): [0,3,3],

    ('D','C','C'): [6,1,1],
    ('D','C','D'): [3,0,3],
    ('D','D','C'): [3,3,0],

    ('D','D','D'): [2,2,2]
}

nash_n_player(strategies, payoffs)
```
